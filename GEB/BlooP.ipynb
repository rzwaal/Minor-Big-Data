```python
def totdemacht_meerdere(grondtal, *exponenten):
    """Berekent het grondtal verheven tot opeenvolgende exponenten.
    Bijvoorbeeld: totdemacht_meerdere(2, 3, 2) berekent (2^3)^2 = 2^(3*2)
    """
    if not exponenten:
        return grondtal

    # Vermenigvuldig alle exponenten met elkaar
    totale_exponent = math.prod(exponenten)
    return grondtal ** totale_exponent
```

```python
totdemacht_meerdere(2,3,2)
```

```python
def totdemacht_rechts_associatief(grondtal, *exponenten):
    """Berekent grondtal tot de macht van de opeenvolgende exponenten van rechts naar links.
    Bijvoorbeeld: totdemacht_rechts_associatief(2, 3, 2) berekent 2^(3^2) = 2^9 = 512
    """
    if not exponenten:
        return grondtal

    # We beginnen met de allerlaatste exponent en werken terug naar voren
    tijdelijke_macht = exponenten[-1]
    for exp in reversed(exponenten[:-1]):
        tijdelijke_macht = exp ** tijdelijke_macht

    return grondtal ** tijdelijke_macht
```

```python
totdemacht_rechts_associatief(2,3,2)
```

## Procedure "MINUS" [M,N]

```python
def minus(m, n):
  if m < n:
    print(m,"is kleiner dan",n,"dus",m,"min",n,"kan niet")
    return

  result = 0
  for i in range(m + 1):
    if result + n == m:
      return result
    result = result + 1
```

```python
minus(5,1)
```

## Procedure "PRIME?" [N]

```python
def prime(n, verbose=True):
  if n <= 1:
    if verbose:
      print(n, "is geen priemgetal")
    return False

  for i in range(2, n):
    if n % i == 0:
      if verbose:
        print(n, "is geen priemgetal, want het is deelbaar door", i)
      return False

  if verbose:
    print(n, "is een priemgetal")
  return True
```

```python
prime(1513689)
```

## Procedure "GOLDBACH?" [N]

https://en.wikipedia.org/wiki/Goldbach%27s_conjecture

```python
def goldbach(n):
  if n % 2 != 0:
    print(n, "is geen even getal")
    return

  if n in [0, 2]:
    print("als het getal gelijk is aan",n,"is er geen oplossing.")
    return

  for i in range(2, n):
    if prime(i, verbose=False) and prime(n - i, verbose=False):
      print(n, "=", i, "+", n - i)
      return
```

```python
goldbach(1513690)
```

## Procedure "FACTORIAL" [N]

```python
def factorial(n):
  if n < 0:
    raise ValueError("Faculteit is niet gedefinieerd voor negatieve getallen.")
  result = 1
  for i in range(1, n + 1):
    result = result * i
  return result
```

## Procedure "PI-DIGIT" [N]

```python
def pi_digit(n):
  if n < 1:
    raise ValueError("n moet 1 of groter zijn om cijfers na de komma te vinden.")

  # Stel de precisie in op n + 10 om afrondingsfouten aan het einde te voorkomen
  decimal.getcontext().prec = n + 10

  # Bereken Pi met behulp van de Chudnovsky-algoritme of een snelle reeks
  # Voor eenvoud en snelheid gebruiken we hier een bekende Leibniz/Machin-achtige formule met decimal
  # Machin-formule: pi/4 = 4*arctan(1/5) - arctan(1/239)
  def arctan(x, precision):
    decimal.getcontext().prec = precision
    x = decimal.Decimal(x)
    result = decimal.Decimal(0)
    term = decimal.Decimal(1) / x
    power = decimal.Decimal(1)
    n_term = 1
    while term != 0:
      if n_term % 2 == 1:
        result += term / power
      else:
        result -= term / power
      power += 2
      term = term / (x * x)
      n_term += 1
    return result

  precision = n + 10
  pi_val = 4 * (4 * arctan(5, precision) - arctan(239, precision))

  # Converteer naar string en haal het n-de cijfer na de komma op
  pi_str = str(pi_val)
  parts = pi_str.split('.')
  if len(parts) > 1 and n <= len(parts[1]):
    return int(parts[1][n - 1])
  else:
    return None
```

```python
# Test de pi_digit functie voor de eerste 5 cijfers na de komma
cijfers = [pi_digit(i) for i in range(1, 16)]
print("De eerste 5 cijfers na de komma zijn:", cijfers)
```

```python
pi_digit(16)
```

## Procedure "FIBO" [N]

```python
def fibo(n):
  if n < 0:
    raise ValueError("Fibonacci is niet gedefinieerd voor negatieve indexen.")
  if n == 0:
    return 0
  if n == 1:
    return 1

  a, b = 0, 1
  for _ in range(2, n + 1):
    a, b = b, a + b
  return b
```

```python
# Test de fibo functie voor de eerste 10 getallen (index 0 tot en met 9)
fib_reeks = [fibo(i) for i in range(10)]
print("De eerste 10 Fibonacci-getallen zijn:", fib_reeks)
```

```python
fibo(10)
```

## Procedure "PRIME-BEYOND" [N]

```python
def prime_beyond(n):
  if n < 2:
    return 2

  candidate = n + 1
  while True:
    if prime(candidate, verbose=False):
      return candidate
    candidate += 1
```

```python
prime_beyond(33)
```

## Procedure "PERFECT_N" [N]

```python
def get_perfect_n(index_or_number):
    # Lijst met de eerste 8 bekende perfecte getallen
    perfect_numbers = [
        6,
        28,
        496,
        8128,
        33550336,
        8589869056,
        137438691328,
        2305843008139952128
    ]

    # Als het invoergetal in de lijst van perfecte getallen staat, geef de 1-gebaseerde index terug
    if index_or_number in perfect_numbers:
        return perfect_numbers.index(index_or_number) + 1

    # Als het invoergetal als index gebruikt kan worden (1-gebaseerd)
    if 1 <= index_or_number <= len(perfect_numbers):
        return perfect_numbers[index_or_number - 1]

    # Foutafhandeling voor ongeldige waarden of getallen buiten ons bereik
    raise ValueError(f"Invoer {index_or_number} is geen bekend perfect getal of een geldige index (1 t/m {len(perfect_numbers)}).")
```

```python
get_perfect_n(2)
```

## Procedure "PERFECT?" [N]

```python
def perfect(n):
  if n <= 0:
    return False

  divisors = [i for i in range(1, n) if n % i == 0]
  print(f"De delers van {n} (kleiner dan {n}) zijn: {divisors}")

  divisors_sum = sum(divisors)
  return divisors_sum == n
```

```python
perfect(496)
```

## Procedure "TRIVIAL?" [A,B,C,N]

```python
def trivial(a, b, c, n):
  if a**n + b**n == c**n:
    return True
  return False
```

## Procedure "PIERRE?" [A,B,C]

```python
def pierre(a, b, c):
  for n in range(2, 100):
    if a**n + b**n == c**n:
      print("De formule klopt voor n =",n)
      return True
  return False
```

## Procedure "FERMAT?" [MAX_VAL, N]

https://en.wikipedia.org/wiki/Fermat%27s_Last_Theorem

```python
def fermat(max_val, n):
  if n <= 2:
    print("De stelling van Fermat gaat over machten groter dan 2. Gebruik n > 2.")
    return False

  print(f"Zoeken naar oplossingen voor a^{n} + b^{n} = c^{n} met a, b, c tot {max_val}...")
  for a in range(1, max_val + 1):
    for b in range(a, max_val + 1):  # b vanaf a om dubbele paren te voorkomen (symmetrie)
      # Bereken wat c^n zou moeten zijn
      c_n = a**n + b**n
      # Bereken de n-de machtswortel van c_n en rond deze af
      c = round(c_n**(1/n))
      if c <= max_val and c**n == c_n:
        print(f"Oplossing gevonden! {a}^{n} + {b}^{n} = {c}^{n} ({a**n} + {b**n} = {c_n})")
        return True

  print(f"Geen oplossingen gevonden voor n = {n} met getallen tot {max_val}.")
  return False
```

```python
fermat(50, 3)
```

## Procedure "TORTOISE-PAIR?" [M,N]

```python
def tortoise_pair(m,n):
  if prime(m) and prime(m+n) is True:
    return True
    return False
```

## Procedure "TORTOISE?" [N]

```python
def tortoise(n):
  # We zoeken naar twee priemgetallen a en b waarvan het verschil n is.
  # Omdat a - b = n, geldt dat a = b + n.
  # We kunnen dit efficiënt testen door priemgetallen te zoeken.
  for b in range(2, 1000): # We nemen een redelijk bereik voor de zoektocht
    a = b + n
    if prime(b, verbose=False) and prime(a, verbose=False):
      print(f"Oplossing gevonden: {a} - {b} = {n}")
      return True
  return False
```

```python
tortoise(7)
```

```python
tortoise(1742)
```
